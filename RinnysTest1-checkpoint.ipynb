```python
import numpy as np
import matplotlib.pyplot as plt

# Fake EEG-like signal
fs = 250  # sampling frequency (Hz)
t = np.arange(0, 2, 1/fs)  # 2 seconds of data
sig = 20e-6*np.sin(2*np.pi*10*t) + 10e-6*np.random.randn(t.size)

plt.figure(figsize=(10,4))
plt.plot(t, sig)
plt.xlabel("Time (s)")
plt.ylabel("Voltage (V)")
plt.title("Sanity check: fake EEG signal")
plt.show()
```
